```python
#UNSUOERVISED LEARNING
#PROJECT: CUSTOMER SEGMENTATION
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
# CUSTOMER DATA
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Annual_Income": [20, 30, 40, 50, 60, 70, 80, 90, 100, 110],
    "Spending_Score": [40, 50, 60, 70, 80, 90, 100, 110, 120, 130]
}
df = pd.DataFrame(data)

#SELECT FEATURES
X = df[["Annual_Income","Spending_Score"]]
#Create K-Mean Model
model = KMeans (n_clusters = 3, random_state =42,n_init = 10)

#Train model
model.fit(X)

#Get cluster number
df["Cluster"] = model.labels_
print("\n Customer Groups:")
print(df)

#Cluster centers
print("CLUSTER CENTERS")
print(model.cluster_centers_)

#VISUALIZATION
plt.scatter(df["Annual_Income"],df["Spending_Score"],c=df["Cluster"],cmap="rainbow")
#plt.scatter(modle.cluster_centers_[;,0],model.cluster_centers_[:,1],color="black")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```

```python

# TRAINING ROBOT TO WALK
import gymnasium as gym # library to create reinforcement-learning environment
from gymnasium import spaces #what action agent can take, what observations the agent can see
import numpy as np #Used to create Robot's state array
from stable_baselines3 import PPO #proximal policy optimization(algorithm) -> PPO is the actual
                                  #algorithm that learns which actions the robot should take

class WalkingRobot(gym.Env): #create the robot environment(state,actions,rewards, goal)
  def __init__(self):
    super().__init__() # calls the constructor of the parent class

    #ACTIONS:
    # 0 -> MOVE LEFT LEG
    # 1 -> MOVE RIGHT LEG
    # 2 -> MOVE BOTH LEGS
    self.action_space = spaces.Discrete(3) #three possible actions

    #ROBOT STATE
    #[left_leg, right_leg, position]
    # what information the robot can observe
    self.observation_space = spaces.Box(low=-10,high=10,shape=(3,),dtype=np.float32)
    self.position=0
    self.left_leg=0
    self.right_leg=0

  def reset(self,seed=None,options=None): #starts a new episode(one attempt by the robot to reach the goal)
    '''episode 1
       Robot starts-> tries action -> reaches goal

       episode 2
       Robot starts again-> tries action -> reaches goal

       episode 3
       Robot starts-> tries action -> reaches goal
    '''

    super().reset(seed=seed) #calls gymnasuiums's reset functionality- seed allows to control randomness
    self.position=0 #reset position to the starting point
    self.left_leg=0 #reset left leg
    self.right_leg=0 #reset right leg

    state = np.array([self.left_leg,self.right_leg,self.position],dtype=np.float32) #creates current state
    return state,{} #state=>[0,0,0], info-> {}

  def step(self,action):
    #Robot moves legs
    if action == 0:
      self.left_leg = 1
      self.right_leg = 0
      self.position += 1

    elif action == 1:
      self.left_leg = 0
      self.right_leg = 1
      self.position += 1

    else:
      self.left_leg = 1
      self.right_leg = 1
      self.position += 2

    #reward for moving forward
    reward = self.position

    #Goal
    terminated = self.position >=20

    #Maximum Steps
    truncated = False

    #create new state
    state = np.array([self.left_leg,self.right_leg,self.position],dtype=np.float32) #   [0,0,5] -> [0,1,6]
    info = {}


    return state,reward,terminated,truncated,{} # [0,1,6],6,False,False,{}

#CREATE ENVIRONMENT
env = WalkingRobot()

#Create PPO Agent
model = PPO("MlpPolicy",env,verbose=1)

#Train Agent/robot
model.learn(total_timesteps=10000)

#Test Robot
state,_ = env.reset()

for step in range(20):
  action, _ = model.predict(state)
  state,reward,done,truncated,_ = env.step(action)
  print(
      "Step:",step,
      "Action:",action,
      "Position:",env.position
  )
  if done:
    print("Robot reached the goal!")
    break
```

```python
import pandas as pd #data manipulation and anaylsis
import numpy as np #numerical operations

import matplotlib.pyplot as plt  #visualization
import seaborn as sns #advanced visualization
from sklearn.preprocessing import StandardScaler #It is used to standardize numerical features before clustering.
from sklearn.cluster import KMeans #K-Means is an unsupervised machine learning algorithm.
from sklearn.metrics import silhouette_score #It helps us determine how well our clusters are separated.
from sklearn.decomposition import PCA #Principal Component Analysis

#np.random.seed(475)

data={
    "Customer_ID":range(1,301),
    "Age":np.random.randint(18,65,300),
    "Annual_Income":np.random.randint(20000,150000,300),
    "Spending_Score":np.random.randint(1,101,300),

    "Purchase_Frequency":np.random.randint(1,30,300),

    "Average_Order_Value":np.random.randint(500,10000,300),

    "Website_Visits":np.random.randint(1,50,300),

    "Discount_Usage":np.random.randint(0, 100, 300)
}

df= pd.DataFrame(data)

df.head()
```

```python
features = [
    "Age",
    "Annual_Income",
    "Spending_Score",
    "Purchase_Frequency",
    "Average_Order_Value",
    "Website_Visits",
    "Discount_Usage"
]

X = df[features]

X.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=features
)

X_scaled_df.head()
```

#find optimal number of clusters

```python
inertia = []

for k in range(2, 11):   # range of K where first k is 2 and,
    model = KMeans( #we are creating kmeans class and its the main algorithym
        n_clusters=k, #n cluster will be in multi groups.

        random_state=42,  # to fix the data order and not shuffle the data
        n_init=10     # (2-10) which is 8 , thrn 8 x 10 = 80 times
    )

    model.fit(X_scaled) # used to train the machine with the  data

    inertia.append(model.inertia_) #inertia_ measures the total squared distance between observations and their assigned cluster centers.
                                   #Lower inertia generally means tighter clusters.But simply increasing K will always tend to reduce inertia, which is why we use the Elbow Method.
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.show()
```

```python
silhouette_scores = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled) # It trains the model and returns the cluster assigned to every customer.

    score = silhouette_score(     #Calculates the Silhouette Score,it evaluates how well each point fits within its cluster compared with neighboring clusters.
        X_scaled,
        labels
    )

    silhouette_scores.append(score)   # this data stores the score.
#Closer to +1 → better separation
 #Around 0     → overlapping clusters
#Below 0      → potentially poor clustering
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.title("Silhouette Score Analysis")

plt.show()
```

```python
best_k = range(2, 11)[
    np.argmax(silhouette_scores)  #argmax is a function just like numpay. #argmax retunrs the position of the normal data
]

print("Best K:", best_k)
```

```python
#Train final Kmeam model

kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

#Add cluster labels to our original dataset:
df["Cluster"] = clusters

df.head()

print(
#Check Number of Customers in Each Cluster
df["Cluster"].value_counts().sort_index()
)
```
